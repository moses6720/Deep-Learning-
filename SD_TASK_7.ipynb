Task 7  build a ResNet-style bottleneck block using PyTorch, with grouped/depthwise convolution, and study the model using the profiler.

Import Libraries

```python
import torch
import torch.nn as nn
```

```python
X = torch.randn(1, 64, 32, 32)

print("Input Shape:", X.shape)
```

Define Bottleneck Block

```python
class Bottleneck(nn.Module):

    def __init__(self):
        super().__init__()

        self.conv1 = nn.Conv2d(64, 32, kernel_size=1)
        self.conv2 = nn.Conv2d(32, 32, kernel_size=3, padding=1, groups=32)
        self.conv3 = nn.Conv2d(32, 64, kernel_size=1)

        self.relu = nn.ReLU()

    def forward(self, x):

        identity = x

        out = self.relu(self.conv1(x))
        out = self.relu(self.conv2(out))
        out = self.conv3(out)

        out = out + identity
        out = self.relu(out)

        return out
```

Create Model

```python
model = Bottleneck()

print(model)
```

Forward Pass

```python
output = model(X)

print("Output Shape:", output.shape)
```

Count parameters

```python
parameters = sum(p.numel() for p in model.parameters())

print("Total Parameters:", parameters)
```
