Task 9: Anchor-Free Object Detection (YOLOv8-style) Loss Backpropagation 

```python
import torch
import torch.nn.functional as F
```

```python
pred_box = torch.tensor([10., 20., 50., 60.])
```

```python
true_box = torch.tensor([15., 25., 55., 65.])
```

```python
def iou(box1, box2):

    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])
    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])

    inter = max(0, x2-x1) * max(0, y2-y1)

    area1 = (box1[2]-box1[0]) * (box1[3]-box1[1])
    area2 = (box2[2]-box2[0]) * (box2[3]-box2[1])

    union = area1 + area2 - inter

    return inter / union

iou_score = iou(pred_box, true_box)
```

```python
ciou_loss = 1 - iou_score
```

```python
pred_class = torch.tensor([0.8])
true_class = torch.tensor([1.0])
class_loss = F.binary_cross_entropy(pred_class, true_class)
```

```python
print("IoU :", iou_score.item())
print("CIoU Loss :", ciou_loss.item())
print("Classification Loss :", class_loss.item())
```
