Task 10: Wasserstein GAN with Gradient Penalty (WGAN-GP) for Stable Image Synthesis 

```python
import torch
import torch.nn as nn
```

```python
generator = nn.Sequential(
    nn.Linear(10, 16),
    nn.ReLU(),
    nn.Linear(16, 4)
)
```

```python
critic = nn.Sequential(
    nn.Linear(4, 16),
    nn.ReLU(),
    nn.Linear(16, 1)
)
```

```python
noise = torch.randn(1, 10)
```

```python
fake = generator(noise)
```

```python
real = torch.randn(1, 4)
```

```python
real_score = critic(real)
fake_score = critic(fake)
```

```python
loss = fake_score.mean() - real_score.mean()
```

```python
print("Real Score :", real_score.item())
print("Fake Score :", fake_score.item())
print("WGAN Loss :", loss.item())
```
