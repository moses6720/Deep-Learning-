Task 11: Vision Transformer (ViT) Patch Projection and Multi-Head Self-Attention from Scratch 

```python
import torch
import torch.nn as nn
```

```python
image = torch.randn(1, 3, 8, 8)
```

```python
patches = image.unfold(2, 4, 4).unfold(3, 4, 4)
patches = patches.contiguous().view(1, 4, -1)
```

```python
print("Patch Shape:", patches.shape)
```

```python
projection = nn.Linear(48, 16)
embeddings = projection(patches)
```

```python
attention = nn.MultiheadAttention(embed_dim=16, num_heads=2, batch_first=True)

output, weights = attention(embeddings, embeddings, embeddings)
```

```python
print("Embedding Shape:", embeddings.shape)
print("Output Shape:", output.shape)
```
