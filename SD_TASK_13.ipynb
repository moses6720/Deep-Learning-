Task 13: Temporal Convolutional Networks (TCN) with Dilated Causal 
Convolutions 


```python
import torch
import torch.nn as nn
```

```python
class TCN(nn.Module):
    def __init__(self):
        super().__init__()

        self.conv1 = nn.Conv1d(1, 4, kernel_size=3, dilation=1, padding=2)
        self.relu = nn.ReLU()
        self.conv2 = nn.Conv1d(4, 1, kernel_size=3, dilation=2, padding=4)

    def forward(self, x):
        x = self.relu(self.conv1(x))
        x = self.conv2(x)
        return x
```

```python
x = torch.randn(1, 1, 10)
```

```python
model = TCN()
```

```python
output = model(x)
```

```python
print("Input Shape :", x.shape)
print("Output Shape:", output.shape)
```
