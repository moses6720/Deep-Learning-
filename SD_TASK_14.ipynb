Integration 


```python
import torch
import matplotlib.pyplot as plt
```

```python
def swish(x):
    return x * torch.sigmoid(x)
```

```python
x = torch.linspace(-5, 5, 100)
```

```python
y = swish(x)
```

```python
print("Input:\n", x[:5])
print("Output:\n", y[:5])

```

```python
plt.plot(x.numpy(), y.numpy())
plt.title("Swish Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.show()
```
